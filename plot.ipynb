Exp 1 : Demonstrate visual perception principles by plotting the same dataset using bar,line, and pie charts and identifying the most effective representation.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:/Users/hetka/OneDrive/Desktop/Data Visulization/Data/netflix1.csv")
df.head()
```

```python
year_count = df['release_year'].value_counts().sort_index()

sns.lineplot(x=year_count.index, y=year_count.values)
plt.grid()
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.title("Titles Released Per Year")
plt.show()
```

```python
sns.countplot(data=df, x='type')
plt.xlabel("Type")
plt.ylabel("Count")
plt.title("Movies vs TV Shows Count")
plt.show()
```

```python
country_count = df['country'].value_counts().head(5)

plt.pie(country_count.values, labels=country_count.index, autopct='%1.1f%%')
plt.title("Top 5 Countries by Content")
plt.show()
```
