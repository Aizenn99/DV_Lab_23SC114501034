Exp 2 : Apply colour theory and contrast by visulization a dataset using different colour palettes and evaluating readability.

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("C:/Users/hetka/OneDrive/Desktop/Data Visulization/Data/netflix1.csv")

type_counts = df['type'].value_counts()

plt.figure()
mono_colors = ['#0d3b66', '#3a86ff']  
plt.bar(type_counts.index, type_counts.values, color=mono_colors)
plt.title("Netflix Content Type (Monochromatic Palette)")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()


```

```python

plt.figure()
comp_colors = ['#1f77b4', '#ff7f0e']  
plt.bar(type_counts.index, type_counts.values, color=comp_colors)
plt.title("Netflix Content Type (Complementary Colors)")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()



```

```python

plt.figure()
analogous_colors = ['#2a9d8f', '#52b788'] 
plt.bar(type_counts.index, type_counts.values, color=analogous_colors)
plt.title("Netflix Content Type (Analogous Palette)")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()


```

```python
plt.figure()
diverging_colors = ['#d62828', '#003049']
plt.bar(type_counts.index, type_counts.values, color=diverging_colors)
plt.title("Netflix Content Type (Diverging Palette)")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()
```
